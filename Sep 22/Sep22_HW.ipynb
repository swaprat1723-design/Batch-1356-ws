```python
name="Swaprat"
#Operation 1: All character printing with positive indexing 

print(name[0])
print(name[1])
print(name[2])
print(name[3])
print(name[4])
print(name[5])
print(name[6])

```

```python
#Operation 2: All character printing with negative indexing (Reverse printing)

print(name[-1])
print(name[-2])
print(name[-3])
print(name[-4])
print(name[-5])
print(name[-6])
print(name[-7])
```

```python
# operation 3: Print 1st and last name

print(name[0], name[-7])
print(name[6], name[-1])
```

```python
# Operation 4: Print last 3 character

print(name[-3],name[-2],name[-1])
```

```python
# Operation 5:Print first 3 character 

print(name[0],name[1],name[2])
```

```python
# Operation 5:Print name 3 times

print(name*3) 
```

```python
# Operation 6: Length of string

print(len(name))
```

```python
# Operation 7: Convert string to Upper and Lower case

print(name.upper())
print(name.lower())
```

```python
# Operation 8: Remove extra spaces

name1=" Python      "

print(name1.strip())
```

```python
# Operation 9: Find character in string

string1= "I love India"

print(string1.find("India"))
```

```python
# Operation 10: Replace text

string1= "I love India"

print(string1.replace("India", "Python"))
```
