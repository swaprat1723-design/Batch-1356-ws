```python
# Composite Operator Assignment

#1. Additionn

a = 10
a += 5

print(f"a: {a}")

# same as

b = 10
b =b + 5

print(f"b: {b}")
```

```python
#2. Subtraction
a = 10
a -= 5

print(f"a: {a}")

# same as

b = 10
b =b - 5

print(f"b: {b}")
```

```python
#3. Multiplication
a = 10
a *= 5

print(f"a: {a}")

# same as

b = 10
b =b * 5

print(f"b: {b}")
```

```python
#4. Division
a = 10
a /= 5

print(f"a: {a}")

# same as

b = 10
b =b / 5

print(f"b: {b}")
```

```python
#5. Modulus
a = 10
a %= 3

print(f"a: {a}")

# same as

b = 10
b =b % 3

print(f"b: {b}")
```

```python
#5. Power of
a = 10
a **= 2

print(f"a: {a}")

# same as

b = 10
b =b **2

print(f"b: {b}")
```

```python
# Bitwise Operator Assignment

# 1. Binary AND or &

a = 5
b = 3

print(a & b) # Output: 1

# Calculation Logic:

#   101   → 5
# & 011   → 3
# -----
#   001   → 1


#--------------------------------------------------------------------------------------------------------------------

# 1. Logical AND:

# Not same as:

print(a and b) # Output: 3

# Calculation Logic:

# 5 is True

#3 is True

#Since both are TRUE, the expression returns the 2nd value


# Learning from Excercise: --

# & → Bitwise AND → works on the binary bits

# and → Logical AND → works on the truth value of objects
```

```python
# 2. Binary OR or |

a = 5
b = 3

print(a | b) # Output: 1

# Calculation Logic:

#   101   → 5
# | 011   → 3
# -----
#   111   → 7 (1 + 2+ 4)


#--------------------------------------------------------------------------------------------------------------------

# 2. Logical OR:

# Not same as:

print(a or b) # Output: 3

# Calculation Logic:

# 5 is True

#3 is True

#Since both are TRUE, the expression returns the 1st value


# Learning from Excercise: --

# & → Bitwise OR → works on the binary bits

# and → Logical OR
#  → works on the truth value of objects
```

```python
# 3. Bitwise XOR OR ^

a = 5
b = 3

print(a ^ b) # Output: 6

# Calculation Logic:

#   101   → 5
# ^ 011   → 3
# -----
#   110   → 6 (0 + 2+ 4)


# Learning from Excercise: --

# 0 ^ 0 = 0 (False)

# 1 ^ 1 = 0 (False)

# Rest all are True
```

```python
# 4. Bitwise NOT OR ~

a = 5
print (~a) # Output:  6

# Calculation Logic:

#   101   → 5
# ~ 010   → 2
# -----
#   010   → 2 (0 + 2+ 0)


# But O/p is -6

# Because ~n = -(n + 1)

# Hence ~5 = - (5 + 1) = - 6


# Learning from Excercise: --

# ~ Not only changes 0 to 1 and 1 to  0

```

```python
# 5. Bitwise Left Shift  OR <<

a = 5
print (a << 2) # Output: 20

# Calculation Logic: 



#    00101   → 5
# << 10100   → <<2
# ---------
#   1010   → 20 (16 + 0 + 4+ 0 + 0)


# Calculation:

# 5 * (2^2) =20


# Learning from Excercise: --

# Number will increase when Left Shifted
```

```python
# 6. Bitwise Left Shift  OR <<

a = 5
print (a >> 2) # Output: 1

# Calculation Logic: 



#    00101   → 5
# >> 00001   → >>2
# ---------
#   00001   → 1 (0 + 0 + 0 + 0 + 1)


# Calculation:

# 5 / (2^2) = 1


# Learning from Excercise: --

# Number will decrease when right Shifted
```
