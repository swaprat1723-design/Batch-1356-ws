```python
'''Write a program to count all letters, digits, and special symbols from a given string.'''

str1 = "P@#yn26at^&i5ve"

chars=0
digits=0
symbols=0

for ch in str1:
    if ch.isalpha():
        chars+=1
    elif ch.isdigit():
        digits+=1
    else:
        symbols+=1
print(chars, digits, symbols)

```

```python
'''Given a string, run a loop to calculate the sum and average of the digits that appear in the string. 
Ignore all other characters.'''

str1 = "PYnative29@#8496"

list_num=[]
list_alpha=[]
list_symbol=[]


for ch in str1:
    if ch.isdigit():
        list_num.append(ch)
    elif ch.isalpha():
        list_alpha.append(ch)
    else:
        list_symbol.append(ch)
sum = 0
for i in list_num:
    sum+=int(i)

avg=sum/len(list_num)

print(f"avg is {avg:.2f}, sum is: {sum}")


```

```python
'''Write a function to determine if a string has all unique characters. 
Return True if every character appears only once, otherwise return False.'''

str1 = "Rama"

set1=set(str1)

count=0


if (len(set1)!=len(str1)):
    print(f"{str1} unique: False")
else:
    for i in set1:
        count+=1
        if str1.count(i)>1:
            print(f"{str1} unique: False")
        elif count==len(str1):
            print(f"{str1} unique: True")
```

```python
'''Write a program to find the longest word in a given sentence. If there is a tie, return the first one found.'''

str1 = "The quick brown fox jumps over the lazy dog"

list1=str1.split()

max_length=len(list1[0])

longest_word=[]

for i in list1:
    if len(i)>max_length:
        max_length=len(i)
        longest_word.insert(0,i)

max_len_word="".join(longest_word)

print(max_len_word, ", length: ", max_length)
```
