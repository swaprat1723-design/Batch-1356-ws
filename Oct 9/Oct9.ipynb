```python
# Prime Number program no

user_input=int(input("Enter the number to check prime or not: "))

prime_flag=False

if user_input <0 :
    print(f"Please enter a positive integer.")

elif user_input ==1:
    print(f"Entered number {user_input} is not a prime number.")

elif user_input>1:
    for i in range(2, user_input):
        if user_input%i==0:
            prime_flag=True
            break

if prime_flag:
    print(f"Entered number {user_input} is not a prime number")
else:
    print(f"Entered number {user_input} is a prime number") 
```
